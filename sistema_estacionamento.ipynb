*SISTEMA PARA ESTACIONAMENTO*


primeira etapa o resgisto de veiculo


```python
print ('Adicione as informações do veiculo:')
lista_mensalista = []
lista_avulso = []

#sistema de laço/reptição
while(True):
  print('-\n NOVO CADASTRO DE CLIENTE')

#informações do cliente
  pergunta=str(input("voce é mensalista (M), avulso (A), sair (S)")).upper()

#Encerramento do laço caso deseje sair  do programa
  if pergunta == 'S':
      print('saindo do sistema')
      break


  Placa = str(input('PLACA'))
  Marca = str(input('MARCA'))
  Modelo = str(input('MODELO'))
  Cor = str(input('COR'))

  valor_hora= 10

  entrada= int(input('horario de entrada'))
  saida= int(input('horario de saida'))

  hora= saida - entrada

  #sistema de cobrança mensalista 12/24 e 24/24
  if pergunta == 'M':
    men_condi = str(input('mensalista integral(I) ou meio periodo(MP)')).upper()

  if men_condi == 'I':
    print('Integral, valor a ser cobrado R$400,00')
    print('Emissão do boleto dia 1 vencimento dia 5')

  elif men_condi == 'MP':
    print('Meio periodo, valor a ser cobrado R$200,00')
    print('Emissão do boleto dia 1 vencimento dia 5')


  #sistema de cobrança do avulso conforme o horario
  if pergunta == 'A':

    if hora <= 1:
      valor1 = 15
      print(f'valor total R${valor1}')

    elif hora <= 3:
      valor2 = 30
      print(f'valor total R${valor2}')

    elif hora > 3:
      valor3 = 30 + (hora - 3) * 5
      print (f'valor total R${valor3}')


  #banco de dados
  if pergunta == 'A':
    dados_cliente = (f'{Placa} | {Marca} | {Modelo} | {Cor}')
    lista_mensalista.append(dados_cliente)
    with open ('mensalista.txt','a') as arquivo:
      arquivo.write(dados_cliente + '\n')
  else:
    dados_cliente = (f'{Placa} | {Marca} | {Modelo} | {Cor}')
    lista_avulso.append(dados_cliente)
    with open ('avulso.txt','a') as arquivo:
      arquivo.write(dados_cliente + '\n')


```
